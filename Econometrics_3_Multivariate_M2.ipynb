# 1. Why add variables?

In Sessions 1–2, we studied the simple regression model:

$$
y_i = \beta x_i + \varepsilon_i,
$$

We learned how to estimate $\beta$, how to measure the uncertainty surrounding $\hat{\beta}$, and how to conduct statistical inference, but most economic and financial relationships involve several factors at the same time.

For example:
* a stock's return may depend on the market, firm size, momentum, and other characteristics;
* a CEO's compensation may depend on firm performance, firm size, industry, and other factors;
* a firm's investment may depend on its expected profitability, its cash flow, its size, and its financing conditions.

**What happens if we want to measure the effect of one variable, but another relevant variable also affects the outcome?**

The problem is not simply that we are "missing an interesting control". If the omitted variable is related to the variable we are interested in, **omitting it can change the coefficient we estimate**. This is the first reason why we need multivariate regression.


*When one variable has information from another*

For simplicity, we start with a model passing through the origin as in Sessions 1-2. We'll get back to the y-intercept later. For observation $i$:
$$
y_i = \beta_1 x_{1i} + \beta_2 x_{2i} + \varepsilon_i,
$$

where:
* $y_i$ is the dependent variable and $x_{1i}$ and $x_{2i}$ are the two explanatory variables,
* $\beta_1$ and $\beta_2$ are the true population parameters,
* $\varepsilon_i$ represents the other factors affecting $y_i$ that are not included explicitly in the model.

Our main assumption is:
$$
\mathbb{E}[\varepsilon_i \mid x_{1i},x_{2i}] = 0.
$$

This notation may look complicated, but the idea is simple: the average value of the error for observations with given values of $x_{1i}$ and $x_{2i}$ is zero. It says that, once we know $x_{1i}$ and $x_{2i}$, the remaining unobserved component $\varepsilon_i$ has an average value of zero. In other words, the error does not systematically increase or decrease with $x_1$ or $x_2$ once both variables are taken into account.

This is the multivariate version of the exogeneity condition we used in the simple regression.

Now suppose that, while the true model is $y_i = \beta_1 x_{1i} + \beta_2 x_{2i} + \varepsilon_i$, we only estimate:
$$
y_i = \tilde{\beta}_1 x_{1i} + \varepsilon_i.
$$

Why we might do this? Perhaps $x_2$ is not available in our dataset or we did not think that $x_2$ was important. Or we simply want to start with a simple model. Whatever the reason, we are now asking the simple regression of $y$ on $x_1$ to measure the effect of $x_1$ **without explicitly accounting for $x_2$**.

Using the simple-regression formula from Session 2, we have:

$$
\tilde{\beta}_1=\frac{\sum_i x_{1i}y_i}
{\sum_i x_{1i}^2}.
$$

We deliberately use the notation $\tilde{\beta}_1$ rather than $\hat{\beta}_1$ to remind us that this coefficient comes from an incomplete specification: the true model contains $x_2$, but our estimated model does not. The question is now: **What does this coefficient actually estimate?**

We know the true model:

$$
y_i = \beta_1 x_{1i} + \beta_2 x_{2i} + \varepsilon_i.
$$

We then substitute this expression for $y_i$ into the formula for $\tilde{\beta}_1$ so that we can simply to express the coefficient we estimate in terms of the three components of the true model:

$$
\tilde{\beta}_1
=
\frac{\sum_i x_{1i}
\left(
\beta_1x_{1i}
+\beta_2x_{2i}
+\varepsilon_i
\right)}
{\sum_i x_{1i}^2}
=
\beta_1
+
\beta_2
\frac{\sum_i x_{1i}x_{2i}}
{\sum_i x_{1i}^2}
+
\frac{\sum_i x_{1i}\varepsilon_i}
{\sum_i x_{1i}^2}.
$$

At this point, we can already see something important. The coefficient obtained from the incomplete regression consists of:

1. the true effect $\beta_1$;
2. an additional term involving $x_2$;
3. a term involving the unobserved error.

We now take expectations. Under the exogeneity assumption
$
\mathbb{E}[\varepsilon_i\mid x_{1i},x_{2i}]=0,
$
the last term has expectation zero. Therefore:

$$
\boxed{
\mathbb{E}[\tilde{\beta}_1]
=
\beta_1
+
\beta_2\delta_1
}
\quad
\text{where}
\quad
\delta_1
=
\frac{\sum_i x_{1i}x_{2i}}
{\sum_i x_{1i}^2}.
$$

$\delta_1$ is precisely the slope coefficient obtained by regressing $x_2$ on $x_1$ in the corresponding simple regression through the origin, and $\mathbb{E}[\tilde{\beta}_1]-\beta_1=\beta_2\delta_1$ is the **omitted variable bias**. The simple-regression coefficient $\tilde\beta_1$ is an unbiased estimator of $\beta_1$ **only if** at least one of two things holds:
- $\beta_2=0$ (the omitted variable $x_2$ truly has no effect on $y$), or
- $\delta_1=0$ ($x_1$ and $x_2$ are uncorrelated in the data).

If the omitted variable matters ($\beta_2\neq0$) **and** is correlated with the variable we kept ($\delta_1\neq0$), then omitting it contaminates our estimate of the effect we actually wanted, $\beta_1$. The sign of the bias is simply the sign of $\beta_2\delta_1$.

This single result is the reason multivariate regression - not simple regression - is the workhorse tool of empirical economics and finance: whenever we suspect a confounding factor (correlated with our variable of interest *and* relevant for the outcome), we must control for it explicitly, by putting it in $\mathbf{X}$.


*A simple example*

Suppose we want to estimate the effect of education on wages:
$$
wage_i = \beta_1 education_i + \beta_2 ability_i + \varepsilon_i.
$$

Suppose ability is difficult to observe, but more able individuals tend to obtain more education.

Then:

* $\beta_2>0$: ability increases wages;
* $\delta_1>0$: ability is positively related to education.

If we regress wages only on education, the education coefficient captures not only the effect of education itself, but also part of the effect of ability.

The estimated coefficient on education is therefore biased upward:

$$
\mathbb{E}[\tilde{\beta}_1]-\beta_1=\beta_2\delta_1>0.
$$

The problem is not that ability is merely an "interesting additional variable", but that **ability affects wages and is correlated with education**. This is why adding relevant explanatory variables can be essential when our objective is to interpret a coefficient as the effect of a particular variable.

As a conclusion, we can say that the coefficient $\beta_1$ is intended to capture the relationship between $x_1$ and $y$ **holding $x_2$ constant**. In a simple regression, the coefficient on $x_1$ reflects the relationship between $x_1$ and $y$ without explicitly accounting for other explanatory variables. In a multivariate regression, we ask a more refined question:

> **How does $y$ differ with $x_1$ among observations that are comparable in terms of the other variables included in the model?**

We are not simply adding more information to the regression. We are trying to **separate the variation associated with different explanatory factors**.

A variable is useful as a control when it helps us separate the variation associated with the variable of interest from variation associated with other factors. But adding variables cannot automatically establish a causal relationship, and including an inappropriate variable can create other problems that we will discuss later.


*What we do next*

We have identified the main motivation for multivariate OLS: When another relevant variable is related to the variable we care about, omitting it can bias our coefficient.

We'll proceed in three steps in the following:

1. **One explanatory variable + a constant:** review the familiar OLS geometry and notation.
2. **Two explanatory variables:** see exactly how adding a second variable changes the coefficient on the first.
3. **The general matrix formulation:** introduce the compact notation that allows us to handle any number of explanatory variables. Introducing matrix notation will give us a clean and general way to express the same OLS problem with several explanatory variables.

# 2. Two variables

##### Adding an intercept

First, we add here an **intercept**. Real data essentially never passes through the origin - there is usually some baseline level:
$$
y_i = \beta_0 + \beta_1 x_i + \varepsilon_i, \qquad i=1,\dots,N.
$$

We now have *two* unknowns, $\beta_0$ and $\beta_1$, instead of one. Let's see that the least-squares logic - minimize the sum of squared residuals, then set derivatives to zero - extends immediately, just with one extra equation to solve.

We write the least-squares program:

$$
(\hat\beta_0,\hat\beta_1) = \underset{\beta_0,\beta_1}{\arg\min}\; \sum_{i=1}^N \big(y_i-\beta_0-\beta_1 x_i\big)^2.
$$

Because there are now two unknowns, minimizing requires setting two partial derivatives to zero - one with respect to $\beta_0$, one with respect to $\beta_1$ (first-order conditions):

$$
\begin{aligned}
&\frac{\partial}{\partial \beta_0}\sum_i(y_i-\beta_0-\beta_1x_i)^2 = -2\sum_i(y_i-\beta_0-\beta_1x_i) = 0\\
&\frac{\partial}{\partial \beta_1}\sum_i(y_i-\beta_0-\beta_1x_i)^2 = -2\sum_i x_i(y_i-\beta_0-\beta_1x_i) = 0.
\end{aligned}
$$

**Setting these two conditions to zero defines the values $\hat\beta_0,\hat\beta_1$ that minimize the sum of squared residuals.** Dividing each by $-2$ and expanding gives the two **normal equations**, now written for this solution:
$$
\begin{aligned}
&\sum_i y_i = N\hat\beta_0 + \hat\beta_1\sum_i x_i, \qquad\qquad\text{(I)}\\
&\sum_i x_iy_i = \hat\beta_0\sum_i x_i + \hat\beta_1\sum_i x_i^2. \qquad\text{(II)}
\end{aligned}
$$

We solve this system by dividing the first equation by $N$, which gives $\bar y=\hat\beta_0+\hat\beta_1\bar x$, i.e.
$$
\hat\beta_0 = \bar y - \hat\beta_1\bar x.
$$
Substituting this into the second equation and rearranging (this is a short but genuine algebra step - try it yourself) leads to:
$$
\boxed{\;\hat\beta_1 = \frac{\sum_i (x_i-\bar x)(y_i-\bar y)}{\sum_i (x_i-\bar x)^2}, \qquad \hat\beta_0 = \bar y - \hat\beta_1 \bar x.\;}
$$

This is exactly the formula you already *used* in TD 1 without a derivation - you now have it. Notice that $\hat\beta_1$ has the same "shape" as the through-origin estimator from Session 2 ($\sum x_iy_i/\sum x_i^2$), except that both variables are first **centered** around their means. Centering is precisely what accounts for the intercept.

##### A second variable

We now add a *second* regressor: including a second variable is how we control for a possible confounder. The model becomes
$$
y_i = \beta_0 + \beta_1 x_{1i} + \beta_2 x_{2i} + \varepsilon_i.
$$
We now have *three* unknowns: $\beta_0,\beta_1,\beta_2$. Let's push the same method - minimize, then differentiate - one step further, and see what happens to the bookkeeping.

$$
\frac{\partial}{\partial\beta_0}\sum_i(y_i-\beta_0-\beta_1x_{1i}-\beta_2x_{2i})^2=0,
$$
$$
\frac{\partial}{\partial\beta_1}\sum_i(y_i-\beta_0-\beta_1x_{1i}-\beta_2x_{2i})^2=0,
$$
$$
\frac{\partial}{\partial\beta_2}\sum_i(y_i-\beta_0-\beta_1x_{1i}-\beta_2x_{2i})^2=0,
$$

which, after expanding (exactly as in Section 1, just with one more term to carry along everywhere), give **three normal equations in three unknowns**:

$$
\begin{aligned}
\sum_i y_i &= N\beta_0 + \beta_1\sum_i x_{1i} + \beta_2 \sum_i x_{2i}, \\
\sum_i x_{1i}y_i &= \beta_0\sum_i x_{1i} + \beta_1\sum_i x_{1i}^2 + \beta_2\sum_i x_{1i}x_{2i}, \\
\sum_i x_{2i}y_i &= \beta_0\sum_i x_{2i} + \beta_1\sum_i x_{1i}x_{2i} + \beta_2\sum_i x_{2i}^2.
\end{aligned}
$$

*Why we need a better notation*

This system *can* be solved by hand (elimination/substitution, exactly as in Section 1), but it is already tedious with only two regressors. With $p$ regressors, we would have $p+1$ normal equations in $p+1$ unknowns - the algebra would be unmanageable, even though the *idea* (minimize a sum of squares, set derivatives to zero) never changes.

This is precisely why econometrics switches to **matrix notation** at this point - not as an arbitrary abstraction, but as a compact bookkeeping device for exactly the same first-order conditions, valid for any number of regressors at once. Before deriving the general case, we first set up the matrix tools we'll need.

**A remark for later.** One can show (we will not do so in detail here) that the coefficient $\hat\beta_1$ obtained from this multivariate regression is generally *different* from the simple-regression slope of $y$ on $x_1$ alone - consistent with Section 1. Intuitively, $\hat\beta_1$ measures the relationship between $y$ and $x_1$ *after* both variables have been "purged" of whatever part of them is explained by $x_2$. This is what "holding $x_2$ fixed" means in practice, and we will come back to this interpretation in Section 5.

# 3. The general case

The three-equation system at the end of Section 2 already shows the issue: with $p$ regressors we would face $p+1$ normal equations in $p+1$ unknowns, all built from the same kind of sums, just multiplying out. Matrix notation is nothing more than a compact way of writing exactly that same system, for any $p$ at once - not a new idea, just better bookkeeping. (If some of the matrix manipulations below feel unfamiliar, a refresher is available in the Appendix at the end of this notebook.)

As in Session 1, we organize the $N$ observations into a vector and a matrix:
$$
y=\begin{pmatrix}y_1\\ \vdots\\ y_N\end{pmatrix}, \qquad
\mathbf{X}=\begin{pmatrix}1 & x_{11} & \cdots & x_{1k}\\ \vdots& \vdots & &\vdots \\ 1 & x_{N1} & \cdots & x_{Nk}\end{pmatrix}, \qquad
\beta=\begin{pmatrix}\beta_0\\ \beta_1\\ \vdots\\ \beta_k\end{pmatrix}, \qquad
\varepsilon=\begin{pmatrix}\varepsilon_1\\ \vdots\\ \varepsilon_N\end{pmatrix}.
$$
$\mathbf{X}$ is $N\times p$ with $p=k+1$ columns: a first column of $1$'s for the intercept $\beta_0$, then one column for each of the $k$ explanatory variables. The model for all $N$ observations is then:
$$
y=\mathbf{X}\beta+\varepsilon.
$$

*Please note that only matrices are written in bold. Like scalars, one-dimensional vectors are not written in bold for readability.*

##### The least-squares program in matrix form

As in Sessions 1-2, we define the residual as a function of a candidate $\beta$: for each observation, $\hat\varepsilon_i(\beta)=y_i-\big(\mathbf{X}\beta\big)_i$, or put together in a vector:
$$
\hat\varepsilon(\beta)=y-\mathbf{X}\beta.
$$

The least-squares program is, as before, to make this as small as possible:
$$
\hat\beta = \underset{\beta}{\arg\min}\; \sum_i\hat\varepsilon_i(\beta)^2.
$$

##### Expanding the objective

To solve the least-squares program, let's first expand the sum of squared residuals, using $(\mathbf{AB})^\top=\mathbf{B}^\top\mathbf{A}^\top$:
$$
\sum_i\hat\varepsilon_i(\beta)^2
=\hat\varepsilon(\beta)^\top\hat\varepsilon(\beta)
=(y-\mathbf{X}\beta)^\top (y-\mathbf{X}\beta)
= y^\top y - y^\top\mathbf{X}\beta - \beta^\top\mathbf{X}^\top y + \beta^\top\mathbf{X}^\top\mathbf{X}\beta.
$$
The two middle terms are both scalars and are transposes of one another ($(\beta^\top\mathbf{X}^\top y)^\top=y^\top\mathbf{X}\beta$), so they are in fact **equal** - this is the transpose trick recalled in the Appendix. We can therefore combine them:
$$
\sum_i\hat\varepsilon_i(\beta)^2 = y^\top y - 2\,\beta^\top\mathbf{X}^\top y + \beta^\top\mathbf{X}^\top\mathbf{X}\beta.
$$
This expression now has exactly the two shapes covered by the two differentiation rules recalled in the Appendix: a linear form in $\beta$ (the middle term) and a quadratic form in $\beta$ (the last term where $\mathbf{X}^\top\mathbf{X}$ is indeed symmetric since $(\mathbf{X}^\top\mathbf{X})^\top=\mathbf{X}^\top\mathbf{X}$).

##### The first-order condition

Differentiating with respect to $\beta$ (the first term $y^\top y$ does not depend on $\beta$, so its derivative is $0$):
$$
\frac{\partial}{\partial\beta}\,\sum_i\hat\varepsilon_i(\beta)^2 = -2\,\mathbf{X}^\top y + 2\,\mathbf{X}^\top\mathbf{X}\beta.
$$
Setting this gradient to zero defines the value $\hat\beta$ that minimizes the sum of squared residuals. This gives the **normal equations** - the matrix generalization of equations (I)-(II) and of the three-equation system from Section 2:
$$
\boxed{\;\mathbf{X}^\top\mathbf{X}\,\hat\beta = \mathbf{X}^\top y.\;}
$$

##### Solving for $\hat\beta$

If the $p\times p$ matrix $\mathbf{X}^\top\mathbf{X}$ is invertible (see "Inverting a matrix" in the Appendix), we can multiply both sides on the left by $(\mathbf{X}^\top\mathbf{X})^{-1}$:
$$
\boxed{\;\hat\beta = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top y.\;}
$$

This single formula contains every case we have studied: for $p=1$ with no intercept, $\mathbf{X}$ is the single column $x$, so $\mathbf{X}^\top\mathbf{X}=\sum_i x_i^2$ (a $1\times1$ "matrix") and $\mathbf{X}^\top y=\sum_i x_iy_i$, giving back exactly $\hat\beta=\sum_i x_iy_i/\sum_i x_i^2$ from Session 2.

##### When does this fail? Perfect multicollinearity

The formula requires $\mathbf{X}^\top\mathbf{X}$ to be invertible. As recalled in the Appendix, this fails exactly when one column of $\mathbf{X}$ can be written as an exact linear combination of the others - for instance, if one explanatory variable is a constant multiple of another, or if a set of dummy variables together with the intercept always sum to $1$ (we will see this concretely later in this notebook, the "dummy variable trap"). When this happens, $\hat\beta$ is not merely hard to compute - it is **not uniquely defined**: infinitely many values of $\beta$ achieve the same (minimal) sum of squared residuals $\hat\varepsilon(\beta)^\top\hat\varepsilon(\beta)$. This is called **perfect collinearity**, and it is a failure of *identification*, not just of arithmetic.

# 4. Properties of the estimator $\hat\beta$

As in Session 2, having a formula for $\hat\beta$ is not the end of the story: we must check that it targets the right value (unbiasedness) and quantify how much it fluctuates (its variance) - only now $\hat\beta$ is a **vector** of $p$ coefficients, so "variance" becomes a full $p\times p$ matrix, collecting the variance of each coefficient *and* the covariances between them.

##### Rewriting $\hat\beta$ in terms of the error

We substitute the true model $y=\mathbf{X}\beta+\varepsilon$ into the formula of $\hat\beta$ we have obtained:
$$
\hat\beta = (\mathbf{X}^\top \mathbf{X})^{-1}\mathbf{X}^\top y = (\mathbf{X}^\top \mathbf{X})^{-1}\mathbf{X}^\top (\mathbf{X}\beta+\varepsilon) = (\mathbf{X}^\top \mathbf{X})^{-1}\mathbf{X}^\top\mathbf{X}\beta + (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\varepsilon.
$$
Since $(\mathbf{X}^\top \mathbf{X})^{-1}\mathbf{X}^\top \mathbf{X}=\mathbf{I}_p$, this simplifies to the matrix version of the Session 2 decomposition:
$$
\boxed{\;\hat\beta = \beta + (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\varepsilon.\;}
$$
Exactly as before: the gap between our estimate and the truth is driven entirely by a term involving the unobserved noise $\varepsilon$.

##### Unbiasedness

We use the same zero conditional mean assumption as Session 2, extended to the whole matrix $\mathbf{X}$: 

$$
\mathbb{E}[\varepsilon\mid \mathbf{X}]=0.
$$

Taking the conditional expectation of the boxed equation above, and noting that $(\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top$ is a fixed matrix once we condition on $\mathbf{X}$:
$$
\mathbb{E}[\hat\beta\mid \mathbf{X}] = \beta + (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\,\underbrace{\mathbb{E}[\varepsilon\mid\mathbf{X}]}_{=\, 0} = \beta.
$$
So $\hat\beta$ is unbiased: this holds for **every** coefficient in the vector simultaneously, not just for one variable at a time - a first formal payoff of the matrix approach.

##### The variance-covariance matrix

We add also the usual assumptions on the errors, now written for the whole vector $\varepsilon$: 

$$
\text{Var}(\varepsilon\mid\mathbf{X}) = \sigma^2\mathbf{I}_N.
$$

It means that each error has the same variance $\sigma^2$ - **homoskedasticity** - and errors for different observations are uncorrelated - exactly the independence assumption of Session 2, generalized).

Using the general rule for the variance of a (fixed matrix) $\times$ (random vector), $\text{Var}(\mathbf{A}Z)=\mathbf{A}\,\text{Var}(Z)\,\mathbf{A}^\top$, with $\mathbf{A}=(\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top$ and $Z=\varepsilon$:
$$
\text{Var}(\hat\beta\mid\mathbf{X}) = (\mathbf{X}^\top\mathbf{X})^{-1}\mathbf{X}^\top\,\big(\sigma^2\mathbf{I}_N\big)\,\mathbf{X}(\mathbf{X}^\top\mathbf{X})^{-1} = \sigma^2(\mathbf{X}^\top\mathbf{X})^{-1}\underbrace{\mathbf{X}^\top\mathbf{X}(\mathbf{X}^\top\mathbf{X})^{-1}}_{=\,\mathbf{I}_p}.
$$
So we obtain the matrix generalization of the Session 2 formula:
$$
\boxed{\;\text{Var}(\hat\beta\mid\mathbf{X}) = \sigma^2(\mathbf{X}^\top\mathbf{X})^{-1}.\;}
$$
This is a $p\times p$ matrix: its **diagonal** entries are $\text{Var}(\hat\beta_0), \text{Var}(\hat\beta_1),\dots$ - exactly what we need for standard errors and t-tests (Section 5) - and its **off-diagonal** entries are the covariances between different estimated coefficients. For $p=1$ (no intercept), $\mathbf{X}^\top\mathbf{X}=\sum_ix_i^2$ and we recover exactly $\text{Var}(\hat\beta)=\sigma^2/\sum_ix_i^2$ from Session 2.

##### Multicollinearity: a variance problem, not a bias problem

Section 3 showed that perfect collinearity (one column of $\mathbf{X}$ exactly reproducible from the others) breaks the formula entirely, since $\mathbf{X}^\top\mathbf{X}$ is not invertible. Near (imperfect) collinearity - two or more regressors that are strongly, but not perfectly, correlated - is a milder, very common, and purely practical problem: $\hat\beta$ still exists and is still unbiased, but $\mathbf{X}^\top\mathbf{X}$ is *close to* singular, so its inverse contains very large entries. From the boxed formula, this directly inflates $\text{Var}(\hat\beta\mid\mathbf{X})$: the standard errors become large, the t-statistics become small, and it becomes hard to detect an effect even if it is genuinely there.

**Intuition.** If two regressors move almost together, the data essentially cannot tell their separate effects apart - many different splits of a combined effect between the two coefficients fit the data nearly equally well, hence the imprecision. Unlike omitted variable bias (Section 1), which is a problem of *what you leave out*, multicollinearity is a problem of *what you put in together*.

# 5. Interpretation, inference, and fit

Three shorter questions remain once $\hat\beta$ and its variance-covariance matrix are in hand: what does a coefficient actually mean with several regressors, how do we test it, and how do we judge the overall fit? None of the three needs new machinery - each is a direct application of Sections 3-4.

##### Ceteris paribus: what does a coefficient mean?

Consider the fitted model $\hat y_i = \hat\beta_0+\hat\beta_1x_{1i}+\hat\beta_2x_{2i}$. Taking a partial derivative with respect to $x_1$, treating $x_2$ as fixed:
$$
\frac{\partial \hat y}{\partial x_1} = \hat\beta_1.
$$
$\hat\beta_1$ is the change in (predicted) $y$ from a one-unit increase in $x_1$, **holding $x_2$ fixed** - in economics this is called the *ceteris paribus* effect ("other things equal").

This is exactly the object Section 1 showed a simple regression *cannot* deliver when a relevant, correlated variable is left out: the simple-regression slope mixes together the direct effect of $x_1$ and the indirect effect coming through its correlation with the omitted $x_2$. Multivariate regression is the tool that separates the two, *provided* the relevant confounders are actually included in $\mathbf{X}$. It cannot control for a variable that is not in the data - a limitation to always keep in mind when interpreting a regression coefficient causally.

##### The t-test with several regressors

Session 2's t-test generalizes immediately, using the variance-covariance matrix from Section 4. For each coefficient $\hat\beta_j$, its standard error is the square root of the $j$-th diagonal entry of the *estimated* variance-covariance matrix, $\widehat{\text{Var}}(\hat\beta\mid\mathbf{X})=\hat\sigma^2(\mathbf{X}^\top\mathbf{X})^{-1}$, where (exactly as in Session 2, but now with $p$ estimated parameters)
$$
\hat\sigma^2 = \frac{1}{N-p}\sum_{i=1}^N \hat\varepsilon_i^2, \qquad \hat\varepsilon_i = y_i - x_i^\top\hat\beta.
$$
The t-statistic for testing $H_0:\beta_j=\beta_{0}$ against $H_1:\beta_j\neq\beta_0$ is, exactly as before,
$$
t_j = \frac{\hat\beta_j - \beta_0}{\text{SE}(\hat\beta_j)} \; \sim \; T_{N-p} \quad \text{under } H_0,
$$
with the same decision rule as Session 2: reject $H_0$ if $|t_j|$ exceeds the critical value $t^*$ for $N-p$ degrees of freedom (or if the p-value is below $\alpha$). The only change from the simple-regression case is that $p$ - the number of estimated parameters, hence the degrees of freedom $N-p$ - now counts the intercept *and every regressor*, not just two.

##### Goodness of fit: $R^2$ and adjusted $R^2$

Define the usual decomposition of the total variation in $y$ around its mean:
$$
\underbrace{\sum_i(y_i-\bar y)^2}_{\text{TSS (total)}} = \underbrace{\sum_i(\hat y_i-\bar y)^2}_{\text{ESS (explained)}} + \underbrace{\sum_i \hat\varepsilon_i^2}_{\text{RSS (residual)}}.
$$
The **coefficient of determination** is
$$
R^2 = \frac{ESS}{TSS} = 1-\frac{RSS}{TSS} \in [0,1],
$$
the share of the total variation in $y$ that the regression explains.

**A mechanical problem.** Adding *any* extra regressor to $\mathbf{X}$ - even one that is economically meaningless, or pure noise - can never *increase* RSS (least squares is, by construction, choosing the best possible fit given what is available, and having one more variable to work with is never worse for minimizing the sum of squared residuals). So $R^2$ mechanically rises (or stays exactly flat) every time we add a variable, regardless of whether that variable actually matters. $R^2$ therefore cannot, by itself, be used to decide whether adding a variable is a good idea.

**The fix.** The **adjusted $R^2$** penalizes for the number of regressors:
$$
\bar R^2 = 1-(1-R^2)\,\frac{N-1}{N-p-1},
$$
where here $p$ counts the regressors excluding the intercept (so $N-p-1$ are the residual degrees of freedom). Unlike $R^2$, $\bar R^2$ can *decrease* when an added variable contributes little, making it a (still imperfect, but more honest) tool for comparing specifications with different numbers of regressors.

Does the ROE-pay link differ by sector? And what happens if we're careless with dummy variables?

## 6.1 Loading the data (recap from TD 1)


```python
from wooldridge import load_data
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

ceo = load_data.data('ceosal1')
ceo.head()
```

## 6.2 Computing $\hat\beta$ by hand

Before letting `statsmodels` do the work, let's see the boxed formula from Section 3 acting on real data. We add a second regressor to the TD 1 model: firm size, measured by (log) sales. Bigger firms tend to pay their CEOs more, largely independently of profitability - so leaving `lsales` out risks exactly the omitted-variable problem from Section 1, if `roe` and `lsales` turn out to be correlated in the data.


```python
y = ceo['lsalary'].values.reshape(-1, 1)          # N x 1
X = sm.add_constant(ceo[['roe', 'lsales']].values)  # N x 3: [1, roe, lsales] - the intercept column, then the two regressors

XtX = X.T @ X            # X'X,  a 3x3 matrix
XtX_inv = np.linalg.inv(XtX)   # (X'X)^{-1} - only possible because X'X is invertible (see Section 3)
Xty = X.T @ y             # X'y,  a 3x1 vector

beta_hat = XtX_inv @ Xty  # exactly the boxed formula of Section 3

print("beta_hat (by hand, matrix formula):")
print(beta_hat)
```

## 6.3 Checking against `statsmodels`

`statsmodels` implements exactly the same formula internally (plus some numerical refinements for stability). Two equivalent ways to call it:


```python
# The matrix API: pass X and y directly, exactly as we built them above
model_matrix_api = sm.OLS(y, X).fit()
print(model_matrix_api.params)
```

```python
# The formula API: describe the model with a patsy formula (adds the intercept automatically)
res = smf.ols('lsalary ~ roe + lsales', data=ceo).fit()
print(res.summary())
```

## 6.5 Does the ROE-pay link differ by sector?

TD 1 ended with a bonus question: does the relationship between `roe` and pay look different for financial firms? We can now answer it formally, by adding an **interaction term** $\text{roe}_i\times\text{finance}_i$ - itself just one more column of $\mathbf{X}$, so nothing about Sections 3-4 changes.

## 6.6 A trap: too many dummy variables

`ceosal1` has **four** mutually exclusive sector dummies: `indus`, `finance`, `consprod`, `utility` - by construction, they sum to exactly $1$ for every firm. Section 3 warned that a column of $\mathbf{X}$ that is an exact linear combination of others makes $\mathbf{X}'\mathbf{X}$ non-invertible.

**Before running the cell below, predict:** what do you expect to happen if we include an intercept **and all four** dummies at once?


```python
res_trap = smf.ols('lsalary ~ roe + indus + finance + consprod + utility', data=ceo).fit()
print(res_trap.summary())
```

Look closely at the output: you should see a **warning about a singular / rank-deficient design matrix**, and a note about a near-zero eigenvalue. This is Section 3's warning made concrete: with an intercept (a column of $1$'s) *and* four dummies that always sum to $1$, the intercept column is an exact linear combination of the four dummy columns. `statsmodels` does not crash - internally it falls back to a generalized inverse - but the individual coefficients it reports are **not economically meaningful**: infinitely many other coefficient vectors would fit exactly as well.

**The standard fix - two equivalent options:**
1. Keep the intercept, and drop **one** dummy (it becomes the reference/"base" category).
2. Drop the intercept, and keep **all** dummies (each dummy's coefficient is then a category-specific intercept).


```python
res_fix_a = smf.ols('lsalary ~ roe + indus + finance + consprod', data=ceo).fit()          # (1) intercept, utility = base category
res_fix_b = smf.ols('lsalary ~ roe + indus + finance + consprod + utility - 1', data=ceo).fit()  # (2) no intercept, all 4 dummies

print("R-squared, option (1):", res_fix_a.rsquared)
print("R-squared, option (2):", res_fix_b.rsquared)
print()
print("Option (1) params:\n", res_fix_a.params)
print()
print("Option (2) params:\n", res_fix_b.params)
```

Both options fit the data **identically** ($R^2$ matches to numerical precision) - they are two different, but equivalent, parametrizations of the exact same fitted values.

**Task.** Check by hand: does `Intercept + indus` in option (1) equal the `indus` row in option (2)? Does `Intercept` alone in option (1) equal the `utility` row in option (2) (the dropped, "base" category)? This is exactly the two encodings of a *fixed effect* you will meet again when working with panel data.

## 6.7 No-intercept models, more generally

Beyond the fixed-effects case just seen, dropping the intercept (`-1` in a patsy formula) is occasionally used when economic theory specifically implies $y=0$ when all regressors are $0$ (a through-origin relationship, as in Sessions 1-2). Outside of that specific case, dropping the intercept is rarely a good idea: it forces the regression line through the origin regardless of whether the data supports it, which typically biases every other coefficient in the process.

# Appendix: Matrix calculations

If you have seen matrices in a finance-mathematics course (e.g. writing portfolio variance as $w'\Sigma w$), most of this will be a refresher; what is likely new is the last part - **differentiating** with respect to a vector of unknowns.

##### Transpose of X
The **transpose** $\mathbf{X}^\top$ of the $N\times p$ matrix $\mathbf{X}$ is the $p\times N$ matrix obtained by turning rows into columns. We have these two useful properties:
- $(\mathbf{AB})^\top = \mathbf{B}^\top\mathbf{A}^\top$ (the transpose of a product reverses the order),
- if $c$ is a scalar (a $1\times 1$ "matrix"), then $c^\top=c$. 

In particular, we have $\big(\beta^\top\mathbf{X}^\top y\big)^\top = y^\top\mathbf{X}\beta$ since $\beta^\top\mathbf{X}^\top y$ is a scalar.

The product $\mathbf{X}^\top\mathbf{X}$ is a $p\times p$ matrix - it collects, in a single object, every sum of products $\sum_i x_{ji}x_{li}$ that appeared one-by-one in the normal equations of Section 2. This is the real payoff of matrix notation: one symbol, $\mathbf{X}^\top\mathbf{X}$, replaces an entire table of sums.

##### Inverting a matrix
For a square ($p\times p$) matrix $\mathbf{A}$, its inverse $\mathbf{A}^{-1}$ - when it exists - is the unique matrix such that $\mathbf{A}^{-1}\mathbf{A}=\mathbf{A}\mathbf{A}^{-1}=\mathbf{I}_p$ (the identity matrix). Just as you cannot divide a number by $0$, $\mathbf{A}^{-1}$ fails to exist when $\mathbf{A}$ is singular (non-invertible) - equivalently, when its determinant is zero, which happens exactly when one column (or row) of $\mathbf{A}$ can be written as a linear combination of the others. We will see below that this is precisely what goes wrong under **perfect multicollinearity**.

##### Quadratic forms
For a $p\times 1$ vector $v$ and a $p\times p$ matrix $\mathbf{A}$, the scalar
$$
v^\top\mathbf{A}v = \sum_{j=1}^p\sum_{l=1}^p A_{jl}\,v_jv_l
$$
is called a **quadratic form** - it generalizes a simple square $v^2$ to several dimensions at once.

You have already met one in finance: portfolio variance $w^\top\Sigma w$ is exactly a quadratic form in the portfolio weights $w$, with $\Sigma$ the covariance matrix. The sum of squared residuals we minimize in this session, $\sum_i\varepsilon_i^2=\varepsilon^\top\varepsilon$, is also a quadratic form (with $\mathbf{A}=\mathbf{I}_N$).

##### Differentiating with respect to a vector

To minimize a function of several unknowns $\beta_0,\dots,\beta_k$ at once, we generalize "take the derivative and set it to zero" to: **take the derivative with respect to *each* component of $\beta$, and stack the results into a vector.** This vector of partial derivatives is called the **gradient**, and is written:

$$
\dfrac{\partial}{\partial\beta}.
$$

Two rules cover everything we need:

**Rule 1 - linear form.** For a fixed $p\times1$ vector $b$:
$$
\frac{\partial}{\partial\beta}\big(b^\top\beta\big) = b.
$$

*Why:* $b^\top\beta=\sum_j b_j\beta_j$ is just a weighted sum, so $\partial(b^\top\beta)/\partial\beta_j=b_j$ for each $j$ - stacking these back into a vector gives exactly $b$.

**Rule 2 - quadratic form.** For a fixed, symmetric $p\times p$ matrix $\mathbf{A}$:
$$
\frac{\partial}{\partial\beta}\big(\beta^\top\mathbf{A}\beta\big) = 2\mathbf{A}\beta.
$$

*A concrete example in dimension 2*

We look at how differentiation works in dimension 2. Take $p=2$:
$$
\mathbf{A}=\begin{pmatrix}a_{11}&a_{12}\\a_{12}&a_{22}\end{pmatrix}
$$

Then:
$$
\beta^\top\mathbf{A}\beta = a_{11}\beta_1^2+2a_{12}\beta_1\beta_2+a_{22}\beta_2^2.
$$
Differentiating with respect to $\beta_1$ gives $2a_{11}\beta_1+2a_{12}\beta_2$, and with respect to $\beta_2$ gives $2a_{12}\beta_1+2a_{22}\beta_2$. Stacking these two into a vector gives exactly $2\mathbf{A}\beta$. The same pattern holds for any $p$, which is the general rule stated above.

With these two rules, we are ready to minimize the sum of squared residuals for **any** number of regressors at once.
